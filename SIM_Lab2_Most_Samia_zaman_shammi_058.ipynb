```python
import numpy as np
from sympy import Matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# My Student ID: 0432320005101058
# Getting d1 and d2
d1 = 8  # last digit
d2 = 5  # second last digit

# 1. Personal Matrix Creation
A = np.array([[d1 + 2, d2 + 1],
              [2 * d1, d2 + 2]])

print("My Matrix A:")
print(A)

# 2. Matrix Analysis
det_A = np.linalg.det(A)
trace_A = np.trace(A)
rank_A = np.linalg.matrix_rank(A)
eigenvalues, eigenvectors = np.linalg.eig(A)

print("\n--- Matrix Properties ---")
print("Determinant:", round(det_A, 2))
print("Trace      :", trace_A)
print("Rank       :", rank_A)

# Checking if inverse exists
if det_A != 0:
    inv_A = np.linalg.inv(A)
    print("\nInverse Matrix:")
    print(np.round(inv_A, 4))
else:
    print("\nInverse does not exist")

print("\nEigenvalues:", np.round(eigenvalues, 4))

# 3. Investigation Part (Changing d1 values)
d1_vals = np.linspace(-10, 10, 100)
det_list = []
trace_list = []

for val in d1_vals:
    temp_mat = np.array([[val + 2, d2 + 1],
                         [2 * val, d2 + 2]])
    det_list.append(np.linalg.det(temp_mat))
    trace_list.append(np.trace(temp_mat))

# Graph plotting
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.plot(d1_vals, det_list, color='blue')
plt.title('d1 vs Determinant')
plt.xlabel('d1')
plt.ylabel('Determinant')
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(d1_vals, trace_list, color='green')
plt.title('d1 vs Trace')
plt.xlabel('d1')
plt.ylabel('Trace')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Student ID: 0432320005101058
# d1 = 8 (last digit), d2 = 5 (second last digit)
d1 = 8
d2 = 5

# Step 1: Matrix A Creation
A = np.array([[d1 + 2, d2 + 1],
              [2 * d1, d2 + 2]], dtype=float)

# Step 2: Analyze Matrix A
shape_A = A.shape
det_A = np.linalg.det(A)
rank_A = np.linalg.matrix_rank(A)
eigenvalues_A, _ = np.linalg.eig(A)

print("Matrix A:")
print(A)

print("\n--- Matrix A Analysis ---")
print("Shape of A:", shape_A)
print("Determinant:", round(det_A, 2))
print("Rank:", rank_A)
print("Eigenvalues:", np.round(eigenvalues_A, 4))

# Inverse computation (only if det != 0)
if det_A != 0:
    inv_A = np.linalg.inv(A)
    print("\nInverse Matrix of A:")
    print(np.round(inv_A, 4))
else:
    print("\nDeterminant is 0, Matrix A has no inverse.")
```

```python
# Step 3: Value Change Experiment

# Copy Matrix A to create Matrix B
B = A.copy()

# Changing exactly one element: Adding +1 to the top-left element (row 0, column 0)
# Matrix A's top-left element was 10, so it becomes 11
B[0, 0] = B[0, 0] + 1

print("Original Matrix A:")
print(A)

print("\nModified Matrix B:")
print(B)

print("\nElement Changed:")
print("Added +1 to element at position A[0, 0] (Row 1, Column 1). Value changed from 10 to 11.")
```

```python
# Step 4: Re-analyze Matrix B

# Computing properties for Matrix B
shape_B = B.shape
det_B = np.linalg.det(B)
rank_B = np.linalg.matrix_rank(B)
eigenvalues_B, _ = np.linalg.eig(B)

print("Matrix B:")
print(B)

print("\n--- Matrix B Analysis ---")
print("Shape of B:", shape_B)
print("Determinant:", round(det_B, 2))
print("Rank:", rank_B)
print("Eigenvalues:", np.round(eigenvalues_B, 4))

# Inverse computation (only if det != 0)
if det_B != 0:
    inv_B = np.linalg.inv(B)
    print("\nInverse Matrix of B:")
    print(np.round(inv_B, 4))
else:
    print("\nDeterminant is 0, Matrix B has no inverse.")
```

## 5. Comparison & Explanation

### 1. How did the determinant change and why?
* **Answer:** The determinant increased from **-26** in Matrix A to **-19** in Matrix B.
* **Reason:** For a 2x2 matrix, $\det(M) = (a \cdot d) - (b \cdot c)$.
  * Matrix A: $(10 \times 7) - (6 \times 16) = -26$
  * Matrix B: $(11 \times 7) - (6 \times 16) = -19$
  Increasing the top-left value increased the main diagonal product, making the determinant less negative.

---

### 2. Did the rank change? Explain.
* **Answer:** No, the rank remained **2** for both matrices.
* **Reason:** A 2x2 matrix has full rank (Rank = 2) when its determinant is non-zero. Since neither $\det(A) = -26$ nor $\det(B) = -19$ is zero, both matrices have linearly independent rows.

---

### 3. How did the eigenvalues respond to the value change?
* **Answer:** The eigenvalues shifted upward (from ~18.68 and -1.68 to ~19.34 and -1.34).
* **Reason:** The sum of eigenvalues equals the trace of the matrix. Increasing $A[0,0]$ by +1 increased the trace from 17 to 18, which correspondingly increased the sum of the eigenvalues.

---

### 4. Is B easier or harder to invert than A? Why?
* **Answer:** Both are **equally easy** to invert.
* **Reason:** Both are standard 2x2 non-singular matrices ($\det \neq 0$). They follow the exact same analytical formula ($M^{-1} = \frac{1}{\det(M)} \text{adj}(M)$) and take the same number of steps to solve.
